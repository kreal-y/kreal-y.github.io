## 4. Mean / Median / Mode with pandas

Now that I have a yearly summary of rat sightings, I use `pandas` to compute some basic statistics:

- **Mean**: the average number of rat sightings per year.
- **Median**: the "middle" yearly value when all years are sorted from lowest to highest.
- **Mode**: the most frequent yearly value (there can be more than one mode).

These statistics give a quick sense of what a “typical” year looks like in terms of rat activity.  
Later, I will re-compute the same values using only the Python standard library.


```python
# Convert the date column to datetime and extract the year
df[date_column] = pd.to_datetime(df[date_column], errors="coerce")
df["year"] = df[date_column].dt.year

# Drop rows where the year is missing
df_year = df.dropna(subset=["year"])

# Group by year and count the number of sightings in each year
yearly_counts = df_year.groupby("year").size().reset_index(name="sightings_count")

# Show the first few rows of the yearly summary
yearly_counts.head()
```

```python
# Select the numeric column we want to analyze
value_column = "sightings_count"

counts_series = yearly_counts[value_column]

# Compute summary statistics using pandas
mean_value = counts_series.mean()
median_value = counts_series.median()
mode_values = counts_series.mode()

print("Mean yearly rat sightings:", mean_value)
print("Median yearly rat sightings:", median_value)
print("Mode(s) of yearly rat sightings:", mode_values.tolist())
```

## 5. Mean / Median / Mode the Hard Way (Pure Python)

For this part of the project, I re-computed the statistics using only the Python standard library:

- I used `csv.DictReader` to read the file row by row.
- I extracted the year from the date string and counted how many sightings occurred in each year.
- I implemented my own functions for the mean, median, and mode.

The results from these custom functions are very close to the ones from `pandas`.  
This is a good check that my logic is correct and that I understand what these statistics mean.


```python
import csv


def read_yearly_counts_from_csv(file_path, date_column_name):
    """
    Read the rat sightings CSV file and return a dictionary
    mapping each year to the number of sightings in that year.
    This uses only the Python standard library.
    """
    yearly_counts_dict = {}
    with open(file_path, newline="", encoding="utf-8") as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            date_str = row[date_column_name]
            if not date_str:
                # Skip rows with missing date
                continue

            # A simple way to extract the year is to take the first 4 characters,
            # assuming the date format starts with "YYYY-..."
            year_str = date_str[:4]
            if not year_str.isdigit():
                # Skip rows where the first 4 characters are not a year
                continue

            year = int(year_str)
            yearly_counts_dict[year] = yearly_counts_dict.get(year, 0) + 1

    return yearly_counts_dict


yearly_counts_dict = read_yearly_counts_from_csv(data_file, date_column)
yearly_counts_dict
```

```python
# Extract the yearly counts as a list of numbers
numbers = list(yearly_counts_dict.values())

# Quick sanity check: show the first few values
numbers[:10]
```

```python
def my_mean(values):
    """
    Compute the arithmetic mean of a list of numbers.
    """
    return sum(values) / len(values)


def my_median(values):
    """
    Compute the median of a list of numbers.
    """
    sorted_values = sorted(values)
    n = len(sorted_values)
    mid = n // 2

    if n % 2 == 1:
        # Odd number of values: return the middle one
        return sorted_values[mid]
    else:
        # Even number of values: return the average of the two middle ones
        return (sorted_values[mid - 1] + sorted_values[mid]) / 2


def my_mode(values):
    """
    Compute the mode(s) of a list of numbers.

    Returns a list because there may be more than one value
    with the highest frequency.
    """
    counts = {}
    for value in values:
        counts[value] = counts.get(value, 0) + 1

    max_count = max(counts.values())
    modes = [v for v, c in counts.items() if c == max_count]
    return modes
```

```python
hard_mean = my_mean(numbers)
hard_median = my_median(numbers)
hard_modes = my_mode(numbers)

print("Mean yearly rat sightings (hard way):", hard_mean)
print("Median yearly rat sightings (hard way):", hard_median)
print("Mode(s) yearly rat sightings (hard way):", hard_modes)
```

## 6. Text-based Visualization (Pure Python)

This text-based chart shows the number of rat sightings in NYC for each year.

- Each line corresponds to a **year**.
- The number of `*` characters on each line is proportional to the number of sightings in that year.
- The chart uses a maximum width of 50 characters, so it fits comfortably on a narrow laptop screen.

Even though this is a very simple visualization, it still allows us to:

- See which years had relatively more or fewer rat sightings.
- Notice any obvious spikes or drops in the yearly counts.


```python
# Sort the dictionary items by year so the chart is in order
sorted_items = sorted(yearly_counts_dict.items())

years = [item[0] for item in sorted_items]
values = [item[1] for item in sorted_items]

years[:5], values[:5]
```

```python
def print_text_bar_chart(labels, values, max_width=50):
    """
    Print a simple horizontal bar chart using '*' characters.

    labels: list of labels (e.g., years)
    values: list of numeric values (e.g., counts)
    max_width: maximum number of '*' characters for the largest value
    """
    max_value = max(values)
    if max_value == 0:
        print("No data to display.")
        return

    scale = max_width / max_value

    print("Text-based Chart: Yearly Rat Sightings in NYC")
    print(
        "Each '*' represents approximately {:.1f} sightings.".format(
            max_value / max_width
        )
    )
    print()
    print("Y-axis: Year")
    print("X-axis: Number of rat sightings (represented by '*')")
    print()

    for label, value in zip(labels, values):
        bar_length = int(value * scale)
        bar = "*" * bar_length
        print(f"{label}: {bar} ({value})")
```

```python
print_text_bar_chart(years, values)
```
